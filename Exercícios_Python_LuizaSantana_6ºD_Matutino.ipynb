20. Dada uma matriz (lista de listas), crie sua matriz transposta.

```python
matriz=[
    [1, 2, 3],
    [4, 5, 6]
]

transposta=[]

for coluna in range(len(matriz[0])):
    novaLinha=[]

    for linha in range(len(matriz)):
        novaLinha.append(matriz[linha][coluna])

    transposta.append(novaLinha)

print(transposta)
```

Usei uma lista de listas para formar uma matriz e criei outra lista vazia para armazenar a matriz transposta. Utilizei o 'for in' com uma range e a função 'len()' para percorrer as colunas da matriz, usando 'len(matriz[0])' para descobrir a quantidade de colunas. Depois, usei um segundo 'for in' para percorrer as linhas da matriz. Para cada coluna, ele pega o elemento correspondente em cada linha usando 'matriz[linha][coluna]' e armazena esses elementos na lista 'novaLinha'. Ao terminar de percorrer as linhas, adiciona 'novaLinha' na lista 'transposta'. Dessa forma, as colunas da matriz original se tornam linhas na matriz transposta, que depois é impressa.


21. Implemente a busca binária para encontrar um elemento em uma lista ordenada.

```python
lista=[2,5,7,9,10,16,24,27,33,42]
inicio = 0
fim = len(lista) - 1

try:
  busca=int(input('Digite o número que deseja encontrar a posição na lista: '))

  while inicio <= fim:
      meio=(inicio+fim)//2 # // para descartar a parte decimal do resultado.

      if lista[meio] == busca:
          print(f'O número {busca} foi encontrado no índice de número {meio} da lista.')
          break
      elif lista[meio] < busca:
          inicio=meio+1
      else:
          fim=meio-1
  else:
      print(f'O número {busca} não foi encontrado na lista.')

except ValueError:
  print('Erro no valor, tente novamente!')
```

Criei uma lista em ordem crescente, uma variável para marcar o início da lista e outra para marcar o fim com a função len() -1 já que os índices começam em 0, as duas variáveis facilitam as comparações seguintes no código. Capturei o número que o usuário deseja encontar a posição dentro da lista  e com o while com a condição para romper o laço quando o ínicio for menor ou igual ao fim, criei a variável 'meio' para somar o início e o fim e dividir por 2, descartando a parte decimal do resultado, com o 'if', verifiquei se o número que ocupa o índice que está no meio da lista era igual ao número fornecido pelo usuário, se sim, imprime que o número foi encontrado em determinada posição e quebra o laco, se não, faz outra verificação com o 'elif' se o meio da lista é maior que o número de busca, se sim, a variável 'início' assume o valor do meio +1 para que que a próxima volta verifique o número da direita, se não, a variável 'fim' assume o valor do meio -1 para que na proxima volta verifique o número da esquerda, caso o while percorra a lista inteira e o número de busca não seja encontrado, imprime o aviso de que o número não está na lista, e para tratar um possível erro no valor, usei o 'try' e o 'except' com a exceção de 'ValueError'.

22. Crie uma função para criptografar uma mensagem usando a cifra de César.

```python
def criptografar(mensagem, deslocamento):
    alfabeto='abcdefghijklmnopqrstuvwxyz'
    mensagemCriptografada=''

    for letra in mensagem:

        if letra in alfabeto:
            posicao=alfabeto.index(letra)
            novaPosicao=(posicao+deslocamento) % 26
            mensagemCriptografada += alfabeto[novaPosicao]
        else:
            mensagemCriptografada += letra

    return mensagemCriptografada

# Teste da função
try:
  mensagem=input('Digite uma mensagem para criptografa-la: ').lower()
  deslocamento=int(input('Digite o deslocamento: '))

  resultado=criptografar(mensagem, deslocamento)

  print('Mensagem criptografada:', resultado)
except ValueError:
  print('Erro no valor, tente novamente!')
```

A função de nome 'criptografar' com os parâmetros string para a mensagem e int para o deslocamento, possui duas variáveis, uma determinada como 'alfabeto' para servir de base no deslocamento das letras porteriormente e outra vazia para armazenar a mensagem que será criptografada.
com o 'for in' percorri cada letra da mensagem, e com o 'if', para cada letra da mensagem presente na variável 'alfabeto', capturei o índice dentro da variável e criei a nova posição somando a posição atual com o deslocamento informado pelo usuário divido pro 26 para quando chegar no último elemento da variável, voltar para o primeiro, depois armazenei a nova letra da mensagem na variável que estava vazia, caso o elemento não seja encontra dentro da variável 'alfabeto', um número por exemplo, o elemento se adiciona na mensagem criptografada na mesma posição da mensagem original. Por fim, capturei a mensagem em uma string com a função lower() para deixar todas as letras da mensagem em minúsculo, capturei o deslocamento desejado pelo usuário e imprimi o resultado aplicando a função criada, e para tratar um possível erro no valor, usei o 'try' e o 'except' com a exceção de 'ValueError'.

24. Implemente uma função para validar um CPF brasileiro.

```python
def validarCpf(cpf):
    cpfLimpo=cpf.replace('.', '').replace('-', '')
    soma=0
    multiplicador=10

    if len(cpfLimpo) != 11 or cpfLimpo == cpfLimpo[0] * 11:
        print('CPF inválido!')
        return

    for numero in cpfLimpo[:9]:
        numero=int(numero)
        numero*=multiplicador
        soma+=numero
        multiplicador-=1

    primeiroDigito=(soma * 10) % 11

    if primeiroDigito == 10:
        primeiroDigito=0

    if primeiroDigito != int(cpfLimpo[9]):
        print('CPF inválido!')
        return

    soma=0
    multiplicador=11

    for numero in cpfLimpo[:10]:
        numero=int(numero)
        numero*=multiplicador
        soma+=numero
        multiplicador-=1

    segundoDigito=(soma * 10) % 11

    if segundoDigito == 10:
        segundoDigito=0

    if segundoDigito != int(cpfLimpo[10]):
        print('CPF inválido!')
        return

    print('CPF válido!')

# Teste da função
cpf = input('Digite o CPF: ')
validarCpf(cpf)
```

A função de nome validarCpf com o parâmetro string para o cpf, recebe o cpf digitado que, caso seja inserido com '.' e '-' é 'limpo' com a função replace() que remove esses elementos e armazena o cpf em uma nova variável para gerantir que na verificação seja usado apenas os números, dentro da função temos duas variáveis já definidas para facilitar as operações matemáticas, uma é definada como 'soma' para armzenar a somas dos elementos do cpf e outras é definida como 'multiplicador' para aplicar a regra de verificação de veracidade de um cpf brasileiro.
O primeiro 'if' verifica se o cpf tem tamanho 11 e se não é formado por 11 números iguais, caso seja, imprime que o cpf é inválido.
O primeiro 'for in' percorre cada número do cpf até o nono digito, transformando o número no formato int para permitir a conta, realiza a multiplicação do mesmo com a variável 'multiplicador' (que na primeira etapa é 10) soma o resultado da multiplicação na variável 'soma' e por último subtrai um do multiplicador para multiplicar todos os elementos por uma sequência decrescente de 10 até 2. Com o resultado da soma da multiplicação dos 10 primeiros números do cpf, é feito a divisão por 11 com o operador '%' para capturar o resto da divisão, essa conta é armazenada na variável 'primeiroDigito'.
O segundo 'if' serve para comparar se o resultado da verificação do primeiro digito é igual a 10, se sim, o 'primeiroDigito' assume o  valor 0.
O terceiro 'if' verifica se a variável 'primeiroDigito' é diferente do décimo número do cpf, se sim, imprime que o cpf é inválido.
Após verificar a veracidade do primiero dígito validador do cpf, partimos para o segunda, que segue quase a mesma lógica, pórem antes a variável 'soma' é redefinida para zero e a 'multiplicador' para 11 pois o primeiro dígito que verificamos faz parte da soma e multiplicação agora.
O segundo 'for in' faz o mesmo papel do primeiro mas até o décimo dígito do cpf, e após cheagar no resultado da soma e multiplicação, varifica se é igual a 10, se sim, a variável 'segundoDigito' assume o valor 0, depois verifica se a variável 'segundoDigito' é diferente do décimo primeiro número do cpf, se sim, imprime que o cpf é inválido, se não, passa pelo último 'if' e valida o cpf.


25. Classifique uma lista de números em três categorias: positivos, negativos e zeros.

```python
import random

lista=[]
for n in range(10):
  lista.append(random.randint(-100,100))

classificacao={'Positivos':[],'Negativos':[],'Zeros':[]}

for n in lista:
  if n > 0:
    classificacao['Positivos'].append(n)
  elif n < 0:
    classificacao['Negativos'].append(n)
  else:
    classificacao['Zeros'].append(n)

print(classificacao)
```

Usei a biblioteca 'random' para criar a lista com números aleátorios limitando-a dentro de um 'for in' com a range 10 para armazenar números entre -100 e 100 na lista vazia, deixando o programa dinâmico visto que a cada run a classificação será diferente. Com outro 'for in', percorri cada elemento da lista passando por três condições: a priemira é que se o número for maior que zero, é classificado como positivo e adicionado no dicionário com na chave 'Positivo', a segunda condição é que se o número for menor que zero, é classificado como negativo e adicionado no dicionário com na chave 'Negativo'

```python
def fatorial(n):
    if n < 0:
        print('O número não pode ser negativo!')
        return
    if n == 0 or n == 1:
        return 1
    return n * fatorial(n - 1)

try:
  n=int(input(f'Digite um número inteiro e positivo: '))
  print(f'O fatorial de {n} é {fatorial(n)}')
except ValueError:
  print('Erro no valor! Tente novamente')
```

48. Implemente o algoritmo de ordenação por inserção (Insertion Sort).

```python
import random

lista=[]
for n in range(10):
    lista.append(random.randint(0, 100))

for i in range(1, len(lista)):
    numero=lista[i]
    j = i - 1

    while j >= 0 and lista[j] > numero:
        lista[j + 1] = lista[j]
        j -= 1

    lista[j + 1] = numero

print(lista)
```

Gerei uma lista vazia e usei o 'for in' com 'range()' para adicionar 10 números aleatórios entre 0 e 100 usando 'random.randint()'. Depois, usei outro 'for in' para percorrer a lista a partir do segundo elemento e armazenei o número atual em 'numero'. Criei 'j' para percorrer os elementos anteriores e usei o 'while' para comparar os valores, deslocando os maiores para a direita. Por fim, coloquei 'numero' na posição correta da lista, formando a lista em ordem crescente, e usei o 'print()' para exibir o resultado.
