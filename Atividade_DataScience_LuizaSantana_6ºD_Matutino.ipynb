19. Crie dois arrays de números inteiros aleatórios de tamanho 20 e calcule o
ângulo entre eles (produto escalar normalizado).

```python
import numpy as np

array1=np.random.randint(1, 10, 20)
array2=np.random.randint(1, 10, 20)

produtoEscalar=np.dot(array1, array2) # Calcula o produto escalar

# Calcular os comprimentos dos vetores
comprimentoA=np.sqrt(np.dot(array1, array1))
comprimentoB=np.sqrt(np.dot(array2, array2))

anguloCos=produtoEscalar / (comprimentoA * comprimentoB) # Calcula o cosseno do ângulo (produto escalar normalizado)

# Encontrar o ângulo em radianos e converter para graus
angulo_rad=np.arccos(anguloCos)
angulo_graus=np.degrees(angulo_rad)

print(f"Produto Escalar: {produtoEscalar}")
print(f"Ângulo em graus: {angulo_graus:.2f}°")
```

Dois arrays unidimensionais foram criados com números inteiros aleatórios de 1 a 9, com um total de 20 elementos cada. Depois, foi calculado o produto escalar entre os dois arrays com a função np.dot(). Em seguida, foi calculado o comprimento de cada vetor usando a raiz quadrada do produto escalar do próprio vetor. Com esses valores, foi calculado o cosseno do ângulo através da divisão do produto escalar pelo produto dos comprimentos dos vetores. Por fim, foi utilizada a função np.arccos() para encontrar o ângulo em radianos e np.degrees() para converter o resultado para graus, sendo exibido o produto escalar e o ângulo com duas casas decimais.

1. Calcule a integral da função f(x) = x^2 sin(x) no intervalo de 0 a π.

```python
import numpy as np
from scipy import integrate

# Define a função a ser integrada
def f(x):
    return (x**2) * np.sin(x)

# Calcula a integral de 0 a pi (np.pi)
resultado, erro = integrate.quad(f, 0, np.pi)

print("Resultado da integral:",resultado)
print("Erro estimado:",erro)
```

Uma função f(x) foi criada para representar a expressão x² · sen(x), utilizando np.sin() para calcular o seno. Depois, a função integrate.quad() da biblioteca SciPy foi utilizada para calcular a integral definida de 0 até π, passando a função e os dois limites como parâmetros. O resultado da integral e uma estimativa do erro do cálculo foram armazenados nas variáveis resultado e erro, respectivamente, sendo exibidos ao final.


2.  Resolva a equação diferencial dy/dx = x - y, com y(0) = 1, no intervalo de 0 a 10.

```python
import numpy as np
from scipy.integrate import solve_ivp

# Define a equação diferencial dy/dx = x - y
def f(x, y):
    return x - y

# Define o intervalo de 0 a 10 e a condição inicial y(0) = 1
intervalo=[0, 10]
y0=[1]

# Gera 100 pontos para avaliar a solução no intervalo
x_pontos=np.linspace(0, 10, 100)

solucao=solve_ivp(f, intervalo, y0, t_eval=x_pontos)

print("Valores de x:",solucao.t)
print("Soluções para y:",solucao.y)
```

Uma função foi criada para representar a equação diferencial dy/dx = x - y, depois foi definido o intervalo de 0 a 10 e a condição inicial y(0) = 1. Em seguida, a função linspace() foi utilizada para gerar 100 pontos igualmente espaçados entre 0 e 10. A função solve_ivp() foi utilizada para resolver a equação diferencial, passando como parâmetros a função, o intervalo, a condição inicial e os pontos onde a solução será avaliada. Por fim, foram exibidos os valores de x e as respectivas soluções de y.

3. Encontre as raízes da função f(x) = x^4 - 3x^3 + 2 utilizando SciPy.


```python
import numpy as np
from scipy.optimize import fsolve

def f(x):
    return x**4 - 3*x**3 + 2

# Encontra uma raiz partindo de um palpite inicial (ex: x = 1)
raiz1=fsolve(f, 1.0)
# Encontra outra raiz partindo de outro palpite (ex: x = 3)
raiz2=fsolve(f, 3.0)

print(f"Raiz encontrada próxima de 1: {raiz1}")
print(f"Raiz encontrada próxima de 3: {raiz2}")
```

A primeira raiz foi procurada partindo do valor 1.0 e armazenada em raiz_1, enquanto a segunda foi procurada partindo do valor 3.0 e armazenada em raiz_2. 4. Use a função de otimização de SciPy para maximizar f(x) = -x^2 + 4x + 1.

```python
from scipy.optimize import minimize

# Para maximizar, precisa usar a função negativa
def f_negativa(x):
    return -(x**2 - 4*x - 1)

# Encontra o mínimo da função negativa partindo de x = 0
resultado=minimize(f_negativa, 0)

# Pega o valor de x dentro do array
ponto_maximo=resultado.x[0]
# Inverte o sinal para obter o valor da função original
valor_maximo=-resultado.fun

print(f"O ponto de máximo encontrado é x = {ponto_maximo:.2f}")
print(f"O valor máximo de f(x) é {valor_maximo:.2f}")
```

Uma função foi criada para representar x² - 4x - 1, depois a função minimize() da biblioteca scipy.optimize foi utilizada para encontrar o menor valor da função, partindo de um palpite inicial x = 0. O resultado foi armazenado em resultado, que contém o ponto onde o menor valor foi encontrado e o valor da função nesse ponto. Em seguida, resultado.x[0] foi utilizado para pegar o primeiro valor do array retornado pelo minimize(), armazenando o ponto de mínimo em ponto_minimo, enquanto resultado.fun armazena o valor mínimo encontrado pela função. Por fim, os dois valores foram exibidos utilizando print(), com :.2f para mostrar apenas duas casas decimais.

5. Calcule a transformada de Fourier de um sinal composto de sin(2πt) e
cos(4πt).


```python
import numpy as np
from scipy.fft import fft

# Cria o intervalo de tempo (0 a 1 segundo, com 500 amostras)
t=np.linspace(0, 1, 500)

# Define o sinal composto de sin(2*pi*t) e cos(4*pi*t), O sinal é a soma das duas frequências
sinal=np.sin(2 * np.pi * t) + np.cos(4 * np.pi * t)

# Calcula a transformada de Fourier (FFT)
frequencias=fft(sinal)

print("Transformada de Fourier (primeiros 5 valores):")
print(frequencias[:5])
```

Um intervalo de tempo foi criado com a função np.linspace(), gerando 500 pontos igualmente espaçados entre 0 e 1 segundo. Depois, foi criado um sinal composto pela soma de duas ondas, uma utilizando np.sin() com frequência de 1 Hz e outra utilizando np.cos() com frequência de 2Hz. Em seguida, a função fft() da biblioteca scipy.fft foi utilizada para calcular a Transformada Rápida de Fourier (FFT) do sinal, transformando o sinal do domínio do tempo para o domínio das frequências. O resultado foi armazenado em frequencias, que contém os valores das frequências presentes no sinal. Por fim, frequencias[:5] foi utilizado para selecionar e exibir apenas os 5 primeiros valores da transformada.

6. Encontre a matriz inversa de uma matriz 5x5 de números aleatorios.

Como o produto de uma matriz pela sua inversa deve resultar em uma matriz identidade, a função np.round(identidade, 2) foi utilizada para arredondar os valores para duas casas decimais e facilitar a visualização da verificação.


7. Use a funções de interpolação cúbica de SciPy para interpolar os dados
(0, 0),(1, 2),(2, 3),(3, 5),(4, 4).

```python
import numpy as np
from scipy.interpolate import interp1d

# Define os pontos de dados conhecidos
x=np.array([0, 1, 2, 3])
y=np.array([1, 3, 2, 5])

# Cria uma função de interpolação usando um polinômio de 3º grau
f_cubica=interp1d(x, y, kind='cubic')

x_novos=np.linspace(0, 3, 7) # Gera 7 novos pontos entre 0 e 3

y_novos=f_cubica(x_novos) # Calcula os valores de y para os novos pontos

print("Novos pontos x:",x_novos)
print("Valores interpolados y:",y_novos)
```

Um array x e um array y foram criados com 4 pontos conhecidos, sendo cada valor de x relacionado a um valor correspondente de y. Depois, a função interp1d() da biblioteca SciPy foi utilizada para criar uma função de interpolação, utilizando kind='cubic' para definir um polinômio de 3º grau. Em seguida, a função np.linspace() foi utilizada para gerar 7 novos pontos igualmente espaçados entre 0 e 3, armazenados em x_novos. Esses novos pontos foram passados para a função f_cubica() para calcular valores de y estimados a partir dos pontos conhecidos, armazenando os resultados em y_novos. para x e [0, 1] para y.


```python
import numpy as np
from scipy.integrate import dblquad

def f(y, x):
    return x * y + x**2

# Calcular a integral dupla
# Limites de x: 0 a 2
# Limites de y: 0 a 1
resultado, erro = dblquad(f, 0, 2, 0, 1)

print(f"Resultado da integral dupla: {resultado:.4f}")
print(f"Erro estimado: {erro}")
```

Na definição da função, y aparece primeiro porque a dblquad() espera os argumentos na ordem (y, x), mesmo que matematicamente normalmente escrevamos f(x, y). O resultado da integral e o erro estimado foram armazenados nas variáveis resultado e erro. Por fim, o resultado foi exibido com :.4f, mostrando 4 casas decimais, e o erro estimado também foi exibido.

```python
import matplotlib.pyplot as plt
import numpy as np

categorias = ['A', 'B', 'C']

valores1 = [3, 5, 4]
valores2 = [2, 4, 3]
valores3 = [5, 7, 6]

# Cria as primeiras barras, que ficam na base do gráfico
plt.bar(categorias, valores1, label='Grupo 1')

# bottom coloca o Grupo 2 acima dos valores do Grupo 1
plt.bar(categorias, valores2, bottom=valores1, label='Grupo 2')

# Soma os dois primeiros grupos para colocar o Grupo 3 acima deles
plt.bar(categorias, valores3, bottom=np.array(valores1) + np.array(valores2), label='Grupo 3')

plt.title('Gráfico de Barras Empilhadas')
plt.xlabel('Categorias')
plt.ylabel('Valores')

plt.legend()  # Exibe a legenda dos grupos
plt.show()    # Exibe o gráfico
```

Em seguida, o parâmetro bottom=valores1 foi utilizado para colocar o Grupo 2 acima dos valores do Grupo 1. Para posicionar o Grupo 3, os valores dos Grupos 1 e 2 foram somados utilizando np.array(), fazendo com que o Grupo 3 começasse acima dos dois grupos anteriores. Depois, plt.title() foi utilizado para definir o título, enquanto plt.xlabel() e plt.ylabel() foram utilizados para nomear os eixos. 18. Crie um gráfico de barras com valores positivos e negativos e use diferentes
cores para cada tipo de valor.

Um array com valores positivos e negativos foi criado para representar os resultados de cada mês. Em seguida, plt.bar() criou o gráfico de barras utilizando essas cores. A função plt.axhline() adicionou uma linha horizontal no valor zero para facilitar a identificação entre lucros e prejuízos.
